```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import glob
import re

def parse_param(filepath, pname):
    match = re.search(rf"{pname}_?(\d+\.?\d*)", filepath)
    return float(match.group(1)) if match else np.nan


innerFiles = glob.glob('initialTotalFlux_grid/innerPlume_R_*')
outerFiles = glob.glob('initialTotalFlux_grid/outerPlume_R_*')

records = []
for inner in innerFiles:
    outer = inner.replace('innerPlume', 'outerPlume')

    R_val = parse_param(inner, 'R')
    U_val = parse_param(inner, 'Usg')

    inner_df = pd.read_csv(inner, index_col=0)

    yi_loc, zi_loc, yo_loc, zo_loc = pf.make_solArray(inner, outer)
    zi_loc = np.asarray(zi_loc)

    ui_bp, Mb, Mb_frac, rho_b, d_eff, Ci, Co, Cs, us, beta, beta_T = \
        pf.getBubbleProperties(yi=yi_loc, zi=zi_loc, yo=yo_loc, zo=zo_loc, ambient=ambient)

    ui_csv, bi, Ti, Si, mb, ci, mdot, zi_csv, _ = pf.getVars(inner_df, model="double_plume", c=c, p=p)

    records.append({
        'R': R_val,
        'Usg': U_val,
        'Qsg': 0.5 * np.pi * R_val**2 * U_val,     # diagnostic, matches the saving script's relation
        'Mb_frac_term': Mb_frac[-1],
        'max_ui': np.max(ui_csv),
        'term_ui': ui_csv.iloc[-1],
        'jp': len(np.where(inner_df["peel"] != 0)[0]),
        'd_eff_term': d_eff[-1],
        'Ci_term': Ci[-1],
        'mdot_max': np.max(mdot),
        'mdot_total': np.trapezoid(mdot, zi_csv),
        'reached_surface': int(np.isclose(zi_loc[-1], ambient[0, -1], atol=1.0)),
    })

results_df = pd.DataFrame(records).sort_values(['R', 'Usg']).reset_index(drop=True)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import cmocean.cm as cm
import glob
import re

def parse_param(filepath, pname):
    match = re.search(rf"{pname}_?(\d+\.?\d*)", filepath)
    return float(match.group(1)) if match else np.nan

innerFiles = glob.glob('initialTotalFlux_grid/innerPlume_R_*')

records = []
for inner in innerFiles:
    outer  = inner.replace('innerPlume', 'outerPlume')
    single = inner.replace('innerPlume', 'singlePlume')

    R_val = parse_param(inner, 'R')
    U_val = parse_param(inner, 'Usg')
    Qsg_val = 0.5 * np.pi * R_val**2 * U_val

    inner_df  = pd.read_csv(inner, index_col=0)
    single_df = pd.read_csv(single, index_col=0)

    yi_loc, zi_loc, yo_loc, zo_loc = pf.make_solArray(inner, outer)
    zi_loc = np.asarray(zi_loc)

    ui_dp, bi, Ti, Si, mb, ci, mdot_dp, zi_csv, _ = pf.getVars(inner_df,  model="double_plume", c=c, p=p)
    u_sp,  b,  T,  S, mdot_sp, z_csv  = pf.getVars(single_df, model="single_plume", c=c, p=p)

    jp = len(np.where(inner_df["peel"] != 0)[0])

    records.append({
        'R': R_val,
        'Usg': U_val,
        'Qsg': Qsg_val,
        'jp': jp,
        'max_dp_mdot': np.nanmax(mdot_dp),
        'max_sp_mdot': np.nanmax(mdot_sp),
        'max_dp_u': np.nanmax(ui_dp),
        'max_sp_u': np.nanmax(u_sp),
    })

results_df = pd.DataFrame(records).sort_values('Qsg').reset_index(drop=True)

results_df['mdot_ratio'] = results_df['max_dp_mdot'] / results_df['max_sp_mdot']
results_df['u_ratio']    = results_df['max_dp_u']    / results_df['max_sp_u']
```

```python
R_vals = np.sort(results_df["R"].unique())
Usg_vals = np.sort(results_df["Usg"].unique())

pivot = results_df.pivot(index="Usg", columns="R", values="u_ratio")
pivot = pivot.reindex(index=Usg_vals, columns=R_vals)  # ensure sorted order

R_grid, Usg_grid = np.meshgrid(R_vals, Usg_vals)
ratio_grid = pivot.values   # shape (len(Usg_vals), len(R_vals)), matches R_grid/Usg_grid

def edges_from_centers(centers):
    centers = np.asarray(centers)
    mid = (centers[:-1] + centers[1:]) / 2
    first = centers[0] - (mid[0] - centers[0])
    last = centers[-1] + (centers[-1] - mid[-1])
    return np.concatenate([[first], mid, [last]])

R_edges = edges_from_centers(R_vals)
Usg_edges = edges_from_centers(Usg_vals)

mask = ratio_grid <= 1  # shape (len(Usg_vals), len(R_vals))

fig, ax = plt.subplots(figsize=(7, 5))
mesh = ax.pcolormesh(R_grid, Usg_grid, ratio_grid, vmin =0 , vmax = 5, shading='auto', cmap=cm.matter_r)
fig.colorbar(mesh, ax=ax, label='$U_{sg}$ ratio')
ax.text(9, 14, r"Ratio = 1.0", color = 'white')

# vertical segments: boundary between horizontally-adjacent cells that disagree
for i in range(mask.shape[0]):
    for j in range(mask.shape[1] - 1):
        if mask[i, j] != mask[i, j + 1]:
            ax.plot([R_edges[j + 1], R_edges[j + 1]],
                     [Usg_edges[i], Usg_edges[i + 1]], color='white', linewidth=2)

# horizontal segments: boundary between vertically-adjacent cells that disagree
for i in range(mask.shape[0] - 1):
    for j in range(mask.shape[1]):
        if mask[i, j] != mask[i + 1, j]:
            ax.plot([R_edges[j], R_edges[j + 1]],
                     [Usg_edges[i + 1], Usg_edges[i + 1]], color='white', linewidth=2)

ax.set_xlabel('R [m]')
ax.set_ylabel('$U_{sg}$')
```

```python

g, phi, u_s = 9.81, 0.1, 0.25             # Qg/QT; bubble slip speed [m/s] -> set to your model's value
NBINS = 12
 
df = results_df.copy()
df['Fr']   = df['Usg'] / np.sqrt(phi * g * df['R'])
df['Us_r'] = df['Usg'] / u_s
 
cands = {r'$Q_{sg}$  (∝ buoyancy flux)': 'Qsg',
         r'$Fr = U_{sg}/\sqrt{\phi g R}$': 'Fr',
         r'$U_{sg}/u_s$': 'Us_r'}
ratios = {'u_ratio':    r'$u_{DPM}/u_{SPM}$',
          'mdot_ratio': r'$\dot m_{DPM}/\dot m_{SPM}$'}
 
# Drop failed single-plume runs (non-positive melt) and any non-positive ratios (log axes)
valid = (df['max_sp_mdot'] > 0) & (df['u_ratio'] > 0) & (df['mdot_ratio'] > 0)
print(f"using {valid.sum()} of {len(df)} runs ({(~valid).sum()} excluded)")
d = df[valid].reset_index(drop=True)
 
Rvals  = np.sort(d['R'].unique())
colors = plt.cm.viridis(np.linspace(0, 0.9, len(Rvals)))
 
 
def collapse_r2(x, y, deg=2):
    lx, ly = np.log10(x), np.log10(y)
    p = np.polyfit(lx, ly, deg)
    res = ly - np.polyval(p, lx)
    return 1 - np.sum(res**2) / np.sum((ly - ly.mean())**2), p
 
 
def spread(x, y, R, nbins=NBINS):
    lx, ly = np.log10(x), np.log10(y)
    edges = np.linspace(lx.min(), lx.max(), nbins + 1)
    s = []
    for a, b in zip(edges[:-1], edges[1:]):
        m = (lx >= a) & (lx <= b)
        means = [ly[m & (R == r)].mean() for r in np.unique(R[m])]
        if len(means) > 1:
            s.append(np.std(means))
    return (np.mean(s) / np.std(ly)) if s else np.nan
 
 
fig, ax = plt.subplots(len(ratios), len(cands), figsize=(5 * len(cands), 4.3 * len(ratios)),
                       constrained_layout=True, squeeze=False)
summary = []
for r_i, (ycol, ylab) in enumerate(ratios.items()):
    for c_i, (xlab, xcol) in enumerate(cands.items()):
        a = ax[r_i, c_i]
        x, y, R = d[xcol].values, d[ycol].values, d['R'].values
        for rv, c in zip(Rvals, colors):
            s = R == rv
            peel, nop = s & (d['jp'] > 0).values, s & (d['jp'] == 0).values
            a.scatter(x[nop],  y[nop],  color=c, s=28, label=f'R = {rv:g}')
            a.scatter(x[peel], y[peel], color=c, s=80, marker='*')
        r2, p = collapse_r2(x, y)
        sp = spread(x, y, R)
        xx = np.logspace(np.log10(x.min()), np.log10(x.max()), 100)
        a.plot(xx, 10**np.polyval(p, np.log10(xx)), 'k--', lw=1)
        a.axhline(1, color='0.7', lw=0.8)
        a.set(xscale='log', yscale='log', xlabel=xlab,
              title=f'R²$_{{collapse}}$ = {r2:.2f},  spread = {sp:.2f}')
        if c_i == 0:
            a.set_ylabel(ylab)
        summary.append((ycol, xcol, r2, sp))
ax[0, 0].legend(fontsize=8, ncols=2)
fig.savefig('collapse_ratios.png', dpi=150)
plt.show()
 
print(f"\n{'ratio':12s} {'x':6s} {'R2_collapse':>12s} {'spread':>8s}")
for ycol, xcol, r2, sp in summary:
    print(f"{ycol:12s} {xcol:6s} {r2:12.3f} {sp:8.3f}")
 
# Power-law fit of mdot_ratio against Fr, non-peeling runs only (peeling changes the physics)
m = (d['jp'] == 0).values
b, loga = np.polyfit(np.log10(d.loc[m, 'Fr']), np.log10(d.loc[m, 'mdot_ratio']), 1)
print(f"\nnon-peeling runs: mdot_ratio ≈ {10**loga:.3g} · Fr^{b:.3f}")
 
```
